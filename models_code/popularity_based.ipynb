# Popularity based recommendation system

```python
rate_w_books = rate.merge(book,on="ISBN")
```

```python
rate_w_books = rate_w_books.drop(columns ="User-ID")
```

```python
rate_w_books
```

```python
num_rate_df = rate_w_books.groupby("Book-Title").count()['Book-Rating'].reset_index()
```

```python
num_rate_df.rename(columns ={'Book-Rating' : 'Num-rate'},inplace = True)
num_rate_df
```

```python
avg_rate_df = rate_w_books.groupby("Book-Title")['Book-Rating'].mean().reset_index()
avg_rate_df.rename(columns ={'Book-Rating' : 'Avg-rate'},inplace = True)
avg_rate_df
```

```python
popular_df = num_rate_df.merge(avg_rate_df,on ='Book-Title')
popular_df
```

```python
popular_df=popular_df[popular_df['Num-rate']>=250].sort_values('Avg-rate', ascending = False).head(50)
```

```python
popular_df.columns.tolist()
```

```python
popular_df=popular_df.merge(book,on='Book-Title').drop_duplicates('Book-Title')[['Book-Title' , 'Book-Author' , 'Image-URL-M',  'Publisher', 'Year-Of-Publication', 'Num-rate', 'Avg-rate' ]]
```
