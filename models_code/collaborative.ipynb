```python
book = pd.read_csv('Books.csv')
rate = pd.read_csv('Ratings.csv')
user = pd.read_csv('Users.csv')
```

```python
rate_w_name = rate.merge(book,on ='ISBN')
```

```python
x = rate_w_name.groupby('User-ID').count()['Book-Rating'] > 200
good_users = x[x].index
```

```python
filtered_df = rate_w_name[rate_w_name['User-ID'].isin(good_users)]
```

```python
y = filtered_df.groupby('Book-Title').count()['Book-Rating'] >= 50
```

```python
y[y]
```

```python
famous_books = y[y].index
```

```python
final_rate = filtered_df[filtered_df['Book-Title'].isin(famous_books)]
```

```python
final_rate.duplicated().sum()
```

```python
pt = final_rate.pivot_table(index = 'Book-Title' , columns = 'User-ID', values = 'Book-Rating')
```

```python
pt
```

```python
pt.fillna(0,inplace=True)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_score = cosine_similarity(pt)
```

```python
def recommendation(book_name):
    #index fetch
    index = np.where(pt.index == book_name)[0][0]
    similar_items = sorted(list(enumerate(similarity_score[index])),key = lambda x:x[1], reverse=True)[1:6]

    data =[]
    for i in similar_items:
        item = []
        temp_df = book[book['Book-Title'] == pt.index[i[0]]]
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Title']))
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Book-Author']))
        item.extend(list(temp_df.drop_duplicates('Book-Title')['Image-URL-M']))

        data.append(item)

    return data
  
```

```python
suggestions = recommendation('1984')
```

```python
suggestions
```

```python
import pickle 
pickle.dump(suggestions,open('recommend.pkl','wb'))
```

```python
pickle.dump(pt,open('pt.pkl', 'wb'))
pickle.dump(book,open('book.pkl', 'wb'))
pickle.dump(similarity_score,open('similarity_score.pkl', 'wb'))
```
